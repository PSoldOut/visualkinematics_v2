```python
from pythreejs import *
from ipywidgets import *
from IPython.display import display
import numpy as np
import util
import manipulator
import time
from scipy.spatial.transform import Rotation as R, Slerp
import pythreejs as three


env = util.Environment()
env.light.position = [5,5,25]

#robot = util.Manipulator("crb15000_5_95")
#robot = manipulator.Manipulator("irb52_7_145")
#robot = manipulator.Manipulator("irb120t_3_58")
#robot = manipulator.Manipulator("irb1200_7_70")

#robot = manipulator.Manipulator("kr5_arc")             #läuft vermutlich wegen experimental nicht
#robot = manipulator.Manipulator("kr210l150")         #läuft vermutlich wegen experimental nicht
#robot = manipulator.Manipulator("kr150_2")
#robot = manipulator.Manipulator("lbr_iiwa_14_r820")
robot = manipulator.Manipulator("irb6640_185_280")


env.add([robot])
display(env)
robot.set_opacity(0.5)
#robot.show_DH_frames()




joint_transforms = robot.calc_global_joint_transforms()
#print(len(joint_transforms))
for transform in joint_transforms:
    g = three.Group()
    frame = util.create_axes(len=0.25, show_labels=False, transparent_arrows=False, arrow_size=0.1)
    g.add(frame)
    util.apply_transformation_matrix(frame, transform)
    #env.add(g)

dh_parameters = []
dh_parameters.append({"theta":0,        "d":0.78,   "a":0,      "alpha":0})
dh_parameters.append({"theta":0,        "d":0,      "a":0.32,   "alpha":np.pi/2})
dh_parameters.append({"theta":np.pi/2,  "d":0,      "a":1.075,  "alpha":0})
dh_parameters.append({"theta":0,        "d":0,      "a":0.2,    "alpha":np.pi/2})
dh_parameters.append({"theta":0,        "d":1.392,  "a":0,      "alpha":-np.pi/2})
dh_parameters.append({"theta":0,        "d":0,      "a":0,      "alpha":np.pi/2})
dh_parameters.append({"theta":0,        "d":0.2,    "a":0,      "alpha":0})
DH = manipulator.DHKinematicModel(dh_parameters=dh_parameters)
DH_transforms = DH.compute_transforms()
for transform in DH_transforms:
    g = three.Group()
    frame = util.create_axes(len=0.30, show_labels=False, transparent_arrows=False, arrow_size=0.1)
    g.add(frame)
    util.apply_transformation_matrix(frame, transform)
    #env.add(g)


print(len(joint_transforms))
print(len(DH_transforms))
#T_align = np.linalg.inv(T_urdf) @ T_dh
for i in range(len(joint_transforms)):
    joint = robot.joints[i]
    joint.dh_alignment = np.linalg.inv(joint_transforms[i]) @ DH_transforms[i]


for i in range(len(joint_transforms)):
    joint = robot.joints[i]
    g = three.Group()
    frame = util.create_axes(len=0.30, show_labels=False, transparent_arrows=False, arrow_size=0.1)
    g.add(frame)
    util.apply_transformation_matrix(frame, joint_transforms[i]@joint.dh_alignment)
    env.add(g)
```
