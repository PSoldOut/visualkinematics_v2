```python
from pythreejs import *
from ipywidgets import *
from IPython.display import display
import numpy as np
import util
import time




env = util.Environment()
env.light.position = [5,5,25]

#robot = util.Manipulator("kr210l150")
#robot = util.Manipulator("kr150_2")
robot = util.Manipulator("irb6640_185_280")
env.add_gizmo_controls(robot.get_joint_by_name("joint_piston"))
#env.add_gizmo_controls(robot.joints[2])
#env.add_gizmo_controls(robot.joints[3])


env.add([robot])
display(env)
```

```python
util.apply_joint_angle(robot.joints[0], robot.joints[0].axis, 0)
util.apply_joint_angle(robot.joints[1], robot.joints[1].axis, 0)
util.apply_joint_angle(robot.joints[2], robot.joints[2].axis, 0)
util.apply_joint_angle(robot.joints[3], robot.joints[3].axis, 0)
util.apply_joint_angle(robot.joints[4], robot.joints[4].axis, 0)

i = 0
while(i < 200):
    util.apply_joint_rotation(robot.joints[4], robot.joints[4].axis, -np.pi/600)
    #time.sleep(0.08)
    util.apply_joint_rotation(robot.joints[3], robot.joints[3].axis, np.pi/200)
    #time.sleep(0.08)
    util.apply_joint_rotation(robot.joints[2], robot.joints[2].axis, -np.pi/800)
    #time.sleep(0.08)
    util.apply_joint_rotation(robot.joints[1], robot.joints[1].axis, np.pi/800)
    #time.sleep(0.08)
    util.apply_joint_rotation(robot.joints[0], robot.joints[0].axis, np.pi/400)
    i+=1
    time.sleep(0.02)
```
