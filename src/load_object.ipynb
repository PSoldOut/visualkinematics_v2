```python
from scipy.spatial.transform import Rotation as R
import numpy as np

# Beispiel-Quaternion (x, y, z, w)
q = [0.0, 0.7071, 0.0, 0.7071]  # 90° Rotation um Y-Achse

# Rotation-Objekt aus Quaternion
r = R.from_quat(q)

# Achse und Winkel extrahieren
rotvec = r.as_rotvec()
angle = np.linalg.norm(rotvec)
axis = rotvec / angle if angle != 0 else np.array([0, 0, 1])  # Default-Achse, falls keine Rotation

print("Rotationswinkel (in Radiant):", angle)
print("Rotationsachse:", axis)
```
