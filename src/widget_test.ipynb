```python
import ipywidgets as widgets
from IPython.display import display

# Dummy "renderer"
renderer = widgets.Label("Renderer (z.B. 3D-Szene)")

# Dummy-Widgets zum Scrollen (viele Sliders)
dummy_widgets = [widgets.FloatSlider(description=f"Joint {i}") for i in range(20)]

# Scrollbarer Container für Widgets
scrollable_widget_box = widgets.VBox(dummy_widgets)

scroll_container = widgets.Box(
    [scrollable_widget_box],
    layout=widgets.Layout(
        overflow_y='auto',
        height='300px',
        width='300px',
        border='1px solid gray',
        padding='10px',
        flex='0 0 auto',
        display='block',
    )
)

# Linke Seite: Renderer + evtl. Checkboxen
left_box = widgets.VBox([
    renderer,
    widgets.HBox([
        widgets.Checkbox(description="Show Grid", value=True),
        widgets.Checkbox(description="Show Axes", value=True)
    ])
])

# Gesamtansicht: links Renderer, rechts Widgets mit Scrollbar
layout = widgets.HBox([
    left_box,
    scroll_container
], layout=widgets.Layout(width='100%'))

display(layout)
```
